$$L(X,y,w, \lambda) = \dfrac{1}{N} \| Xw - y \|_2^2 + \lambda \|w\|_2^2$$

В качестве метрики качества здесь и далее предлагается использовать **MAE**.

```python
import itertools
from sklearn.metrics import mean_absolute_error
from modules.linear_model import LinearModel
from modules.losses import LinearLoss
```

```python
def test_with_params(
        X_train, y_train, X_val, y_val,
        alphas: list,
        betas: list,
        batches: list = [None], 
        l2=0, 
        max_iter=500, 
        trace=False):
    '''
    Runs model with all combinationa of parametrs

    Return extended hsitory dict:

    {
        (alpha, beta, batch): 
            {
                "time": [],
                "func": [],
                "func_val": [],
                "mae_train": float,
                "mae_val": float,
            }
    }
    '''
    combinations = list(itertools.product(alphas, betas, batches))
    history = {}
    loss_function = LinearLoss(l2_coef=l2)
    for a, b, batch in combinations:
        m = LinearModel(
            loss_function=loss_function,
            step_alpha=a,
            step_beta=b,
            batch_size=batch,
            max_iter=max_iter)
        h = m.fit(X=X_train, y=y_train, X_val=X_val, y_val=y_val, trace=trace)
        mae_train = mean_absolute_error(y_train, m.predict(X_train)) 
        mae_val = mean_absolute_error(y_val, m.predict(X_val)) 
        if h is None:
            h = {}
        h["mae_train"] = mae_train
        h["mae_val"] = mae_val
        history[(a, b, batch)] = h
    return history
```

```python
history_alphas_betas = test_with_params(
    X_train=X_train, 
    y_train=y_train,
    X_val=X_val,
    y_val=y_val,
    alphas=[0.1, 0.5, 1],
    betas=[0.1, 0.3, 0.5, 0.7],
    trace=True
    )
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 5))
for (a, b, _), h in history_alphas_betas.items():
    sns.lineplot(x=h['time'], y=h['func'], label=f'alpha={a}, beta={b}')
plt.xlabel('Time (s)')
plt.ylabel('Loss')
plt.title('Loss vs. Time (Train)')
plt.legend()
plt.show()

plt.figure(figsize=(12, 5))
for (a, b, _), h in history_alphas_betas.items():
    sns.lineplot(x=h['time'], y=h['func_val'], label=f'alpha={a}, beta={b}')
plt.xlabel('Time (s)')
plt.ylabel('Loss')
plt.title('Loss vs. Time (Val)')
plt.legend()
plt.show()

plt.figure(figsize=(12, 5))
for (a, b, _), h in history_alphas_betas.items():
    sns.lineplot(x=range(len(h['func'])), y=h['func'], label=f'alpha={a}, beta={b}')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.title('Loss vs. Epoch (Train)')
plt.legend()
plt.show()

plt.figure(figsize=(12, 5))
for (a, b, _), h in history_alphas_betas.items():
    sns.lineplot(x=range(len(h['func_val'])), y=h['func_val'], label=f'alpha={a}, beta={b}')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.title('Loss vs. Epoch (Val)')
plt.legend()
plt.show()
```

```python
def to_df(history: dict):
    if len(list(history_alphas_betas.keys())[0]) > 2:
        mae_data = [(a, b, batch, h['mae_train'], h['mae_val']) for (a, b, batch), h in history.items()]
        mae_df = pd.DataFrame(mae_data, columns=['step_alpha', 'step_beta', 'batch_size', 'mae_train', 'mae_val'])
    else:
        mae_data = [(a, b, h['mae_train'], h['mae_val']) for (a, b), h in history.items()]
        mae_df = pd.DataFrame(mae_data, columns=['step_alpha', 'step_beta', 'mae_train', 'mae_val'])
    return mae_df
```

```python
alphas_betas_mae_df = to_df(history_alphas_betas)
```

```python
plt.figure(figsize=(8, 4))
sns.barplot(data=alphas_betas_mae_df, x="step_alpha", y="mae_train", hue="step_beta")
plt.title("MAE (Train) for Different step_alpha and step_beta")
plt.xlabel("Step Alpha")
plt.ylabel("MAE (Train)")
plt.legend(title="Step Beta")
plt.show()

plt.figure(figsize=(8, 4))
sns.barplot(data=alphas_betas_mae_df, x="step_alpha", y="mae_val", hue="step_beta")
plt.title("MAE (Validation) for Different step_alpha and step_beta")
plt.xlabel("Step Alpha")
plt.ylabel("MAE (Validation)")
plt.legend(title="Step Beta")
plt.show()
```

Замечание. Обратите внимание, что в стохастическом случае необходимо строить зависимости метрик качества от эпохи метода. За одну эпоху через оптимизацию модели проходит $N$ объектов, где $N$ — длина обучающей выборки.
